# Task 8: Multi-Agent System

## Objective

Implement a simple multi-agent system where multiple specialized agents
work together to solve a problem.

This task demonstrates:

- Multiple AI agents
- Agent specialization
- Task delegation
- Agent communication
- Sequential workflow
- Final result generation

The system contains three agents:

1. Research Agent
2. Analysis Agent
3. Final Response Agent

```python
print("Multi-Agent System")

print("Libraries loaded successfully!")
```

## Step 1: Research Agent

The Research Agent collects relevant information about the given topic.

In a real-world system, this agent could search websites, databases,
documents, or APIs.

```python
class ResearchAgent:

    def research(self, topic):

        information = {
            "topic": topic,
            "facts": [
                f"{topic} is an important area of technology.",
                f"{topic} is being used in multiple real-world applications.",
                f"{topic} can improve automation and decision making."
            ]
        }

        return information


research_agent = ResearchAgent()

research_result = research_agent.research(
    "Generative AI"
)

print("Research Agent Output:")
print(research_result)
```

## Step 2: Analysis Agent

The Analysis Agent receives the information collected by the Research Agent.

It analyzes the information and identifies the main points.

```python
class AnalysisAgent:

    def analyze(self, research_data):

        facts = research_data["facts"]

        analysis = {
            "topic": research_data["topic"],
            "key_points": facts,
            "number_of_facts": len(facts)
        }

        return analysis


analysis_agent = AnalysisAgent()

analysis_result = analysis_agent.analyze(
    research_result
)

print("Analysis Agent Output:")
print(analysis_result)
```

## Step 3: Final Response Agent

The Final Response Agent receives the analyzed information and produces
a concise final response.

```python
class ResponseAgent:

    def generate(self, analysis_data):

        topic = analysis_data["topic"]
        points = analysis_data["key_points"]

        response = f"Summary about {topic}:\n\n"

        for point in points:

            response += "- " + point + "\n"

        return response


response_agent = ResponseAgent()

final_response = response_agent.generate(
    analysis_result
)

print("Final Response:")
print(final_response)
```

## Step 4: Agent Communication

The agents communicate sequentially.

Research Agent → Analysis Agent → Response Agent

The output of one agent becomes the input of the next agent.

```python
print("Research Agent")
print("      ↓")
print("Analysis Agent")
print("      ↓")
print("Response Agent")
```

## Step 5: Complete Multi-Agent Workflow

We combine all three agents into one workflow.

The user provides a topic, and the agents work together to produce
the final response.

```python
def multi_agent_system(topic):

    # Agent 1: Research
    research_result = research_agent.research(
        topic
    )

    # Agent 2: Analysis
    analysis_result = analysis_agent.analyze(
        research_result
    )

    # Agent 3: Final Response
    final_response = response_agent.generate(
        analysis_result
    )

    return final_response
```

## Step 6: Run the Multi-Agent System

We test the complete system with a topic.

```python
topic = "Artificial Intelligence"

result = multi_agent_system(
    topic
)

print(result)
```

## Step 7: Test with Another Topic

The same agents can work on a different topic without changing the
agent architecture.

```python
topic2 = "Large Language Models"

result2 = multi_agent_system(
    topic2
)

print(result2)
```

## Step 8: Track Agent Execution

We record which agents participated in solving the task.

```python
agent_log = []

agent_log.append(
    "Research Agent completed research"
)

agent_log.append(
    "Analysis Agent analyzed information"
)

agent_log.append(
    "Response Agent generated final response"
)

print("Agent Execution Log:\n")

for log in agent_log:

    print("✓", log)
```

## Step 9: Single-Agent vs Multi-Agent

A single agent performs the entire task.

A multi-agent system divides the task into specialized responsibilities.

```python
comparison = {
    "Single Agent": [
        "One agent performs all tasks",
        "Less specialization",
        "Simple architecture"
    ],

    "Multi-Agent": [
        "Tasks are divided between agents",
        "Each agent has a specific role",
        "Agents can collaborate"
    ]
}

for system, points in comparison.items():

    print("\n" + system)

    for point in points:

        print("-", point)
```
