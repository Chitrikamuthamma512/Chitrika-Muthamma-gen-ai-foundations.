```python
import torch
import torch.nn as nn

class SimpleNet(nn.Module):

    def __init__(self):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(100,256),
            nn.ReLU(),

            nn.Linear(256,128),
            nn.ReLU(),

            nn.Linear(128,10)

        )

    def forward(self,x):

        return self.network(x)
```
