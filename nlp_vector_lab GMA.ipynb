Text Selection: Create a list containing exactly three specific text strings where the
word &quot;light&quot; is used in different contexts:
o Sentence 1: A review where &quot;light&quot; means low-calorie, weight, or texture
(e.g., &quot;This olive oil is wonderful and very light on the stomach.&quot;).
o Sentence 2: A review where &quot;light&quot; means illumination or physical visibility
(e.g., &quot;The packaging was broken and let too much light inside the box.&quot;).
o Sentence 3: Another review where &quot;light&quot; matches the low-calorie or weight
context of Sentence 1 (e.g., &quot;I prefer this brand because it is a light and
healthy snack.&quot;).

```python
light_reviews = df[df["Text"].str.contains("light", case=False, na=False)]

print(light_reviews["Text"].head(50))
```

```python
sentence1 = "This olive oil is wonderful and very light on the stomach."
sentence2 = "The packaging was broken and let too much light inside the box."
sentence3 = "I prefer this brand because it is a light and healthy snack."
```

```python
sentences = [sentence1, sentence2, sentence3]
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()

matrix = vectorizer.fit_transform(sentences)
```

```python
dense_matrix = matrix.toarray()

print(dense_matrix)
```

```python
vec_1 = dense_matrix[0]
vec_2 = dense_matrix[1]
vec_3 = dense_matrix[2]
```

```python
print(vec_1)
print(vec_2)
print(vec_3)
```

```python
import numpy as np

similarity_12 = np.dot(vec_1, vec_2)

similarity_13 = np.dot(vec_1, vec_3)
```

```python
print("Similarity between Sentence 1 and Sentence 2:")
print(similarity_12)

print()

print("Similarity between Sentence 1 and Sentence 3:")
print(similarity_13)
```
