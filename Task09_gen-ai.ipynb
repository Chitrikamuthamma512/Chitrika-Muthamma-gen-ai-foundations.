# Task 9: Vector Streaming

## Objective

Implement a simple vector streaming system where data arrives continuously
and is converted into numerical vectors.

This task demonstrates:

- Streaming data
- Vector generation
- Incremental storage
- Similarity search
- Real-time retrieval

Instead of processing all data at once, vectors are processed as they arrive.

```python
import numpy as np
import time

np.random.seed(42)

print("Libraries imported successfully!")
```

## Step 1: Simulate Incoming Data

In a real application, data could arrive from:

- Sensors
- User activity
- Logs
- Messages
- Documents

Here we simulate data arriving one item at a time.

```python
stream_data = [
    "Generative AI creates new content.",
    "Machine learning learns patterns from data.",
    "Vector databases support similarity search.",
    "RAG retrieves relevant information.",
    "Transformers are widely used in modern AI."
]

print("Number of incoming items:", len(stream_data))
```

## Step 2: Convert Data into Vectors

Each incoming item is converted into a numerical vector.

For this simplified experiment, we generate a fixed-size vector
from the text.

```python
def create_vector(text, dimension=8):

    seed = sum(
        ord(character)
        for character in text
    )

    rng = np.random.default_rng(seed)

    vector = rng.random(dimension)

    return vector
```

## Step 3: Create an Empty Vector Store

The vector store will hold vectors as they arrive.

Unlike batch processing, the store is updated incrementally.

```python
vector_store = []

print("Vector store created.")
```

## Step 4: Stream Incoming Vectors

Each piece of data is processed as it arrives.

The generated vector is immediately added to the vector store.

```python
for item in stream_data:

    vector = create_vector(item)

    vector_store.append({
        "text": item,
        "vector": vector
    })

    print(
        "Added:",
        item
    )

print("\nStreaming completed.")
print(
    "Vectors stored:",
    len(vector_store)
)
```

## Step 5: Inspect Stored Vectors

We check the number and dimensions of the vectors stored in the system.

```python
print(
    "Number of vectors:",
    len(vector_store)
)

print(
    "Vector dimension:",
    len(
        vector_store[0]["vector"]
    )
)
```

## Step 6: Calculate Vector Similarity

Cosine similarity is used to compare vectors.

A higher similarity score means that two vectors are more similar.

```python
def cosine_similarity(a, b):

    return np.dot(a, b) / (
        np.linalg.norm(a)
        * np.linalg.norm(b)
    )
```

## Step 7: Search the Vector Stream

A query is converted into a vector.

The system compares it with the vectors currently stored in the
vector store.

```python
query = "How does vector search work?"

query_vector = create_vector(query)

results = []

for item in vector_store:

    score = cosine_similarity(
        query_vector,
        item["vector"]
    )

    results.append({
        "text": item["text"],
        "score": score
    })

results.sort(
    key=lambda x: x["score"],
    reverse=True
)

print("Query:")
print(query)

print("\nTop Results:\n")

for result in results[:3]:

    print(
        f"Score: {result['score']:.4f}"
    )

    print(
        result["text"]
    )

    print()
```

## Step 8: Add New Data to the Stream

One advantage of streaming systems is that new data can be added
without rebuilding the entire vector store.

```python
new_data = [
    "Large language models process natural language.",
    "Embeddings convert text into numerical representations."
]

for item in new_data:

    vector = create_vector(item)

    vector_store.append({
        "text": item,
        "vector": vector
    })

    print(
        "New vector added:",
        item
    )

print(
    "\nTotal vectors:",
    len(vector_store)
)
```

## Step 9: Search the Updated Vector Store

We perform the same query again after new vectors have arrived.

```python
results = []

for item in vector_store:

    score = cosine_similarity(
        query_vector,
        item["vector"]
    )

    results.append({
        "text": item["text"],
        "score": score
    })

results.sort(
    key=lambda x: x["score"],
    reverse=True
)

print("Updated Search Results:\n")

for result in results[:3]:

    print(
        f"Score: {result['score']:.4f}"
    )

    print(
        result["text"]
    )

    print()
```

## Step 10: Simulate Real-Time Processing

A small delay is introduced to simulate data arriving continuously.

```python
real_time_data = [
    "AI agents can perform tasks.",
    "Fine-tuning adapts models to specific tasks.",
    "Quantization reduces model memory usage."
]

for item in real_time_data:

    print(
        "\nIncoming data:",
        item
    )

    vector = create_vector(item)

    vector_store.append({
        "text": item,
        "vector": vector
    })

    print("Vector generated and stored.")

    time.sleep(1)

print("\nReal-time stream completed.")
```

## Step 11: Vector Store Statistics

We display basic statistics about the streaming vector store.

```python
total_vectors = len(vector_store)

vector_dimension = len(
    vector_store[0]["vector"]
)

print(
    "Total vectors:",
    total_vectors
)

print(
    "Vector dimension:",
    vector_dimension
)

print(
    "Storage status: Active"
)
```

## Step 12: Results

The vector streaming system successfully demonstrates:

- Continuous data arrival
- Vector generation
- Incremental vector storage
- Similarity search
- Adding new vectors
- Searching updated data
- Real-time processing simulation

```python
print("========== TASK 9 RESULTS ==========")

print(
    "Vectors stored:",
    total_vectors
)

print(
    "Vector dimension:",
    vector_dimension
)

print(
    "Streaming status: Completed"
)

print(
    "Similarity search: Working"
)

print("====================================")
```

# Conclusion
